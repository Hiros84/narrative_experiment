```python
import csv
import os
from PIL import Image, ImageDraw, ImageFont
import textwrap

# === 定数設定 ===
INPUT_FN = "input.csv"        # 入力ファイル（例: "data/products.csv"）
OUTPUT_DIR = "output_images"  # 出力フォルダ名

# === 画像設定 ===
IMG_SIZE = (770, 770)
FONT_MAIN_SIZE = 100   # 1行目フォントサイズ
FONT_SUB_SIZE = 48     # 2行目フォントサイズ
BG_COLOR = (255, 255, 255)
TEXT_COLOR = (0, 0, 0)

# === フォントパス ===
# Mac の場合は以下などが使えます。他の環境の場合はパスを調整してください
FONT_PATH_MAIN = "/System/Library/Fonts/ヒラギノ角ゴシック W6.ttc"
FONT_PATH_SUB = "/System/Library/Fonts/ヒラギノ角ゴシック W3.ttc"

# === 出力フォルダ作成 ===
os.makedirs(OUTPUT_DIR, exist_ok=True)

# === CSV読み込みと画像生成 ===
with open(INPUT_FN, newline='', encoding='utf-8') as csvfile:
    reader = csv.DictReader(csvfile)
    for row in reader:
        exist_flag = row.get("存在チェック", "")
        if exist_flag.strip() == "×":
            product_id = row.get("品番", "unknown")
            name = row.get("名前", "")
            filename = f"{product_id}.jpg"
            output_path = os.path.join(OUTPUT_DIR, filename)

            # 画像生成
            img = Image.new("RGB", IMG_SIZE, BG_COLOR)
            draw = ImageDraw.Draw(img)

            # フォント設定
            font_main = ImageFont.truetype(FONT_PATH_MAIN, FONT_MAIN_SIZE)
            font_sub = ImageFont.truetype(FONT_PATH_SUB, FONT_SUB_SIZE)

            # ===== 1行目（品番名） =====
            w_main, h_main = draw.textsize(product_id, font=font_main)
            main_x = (IMG_SIZE[0] - w_main) / 2
            main_y = (IMG_SIZE[1] - h_main) / 2 - 50  # 少し上に配置
            draw.text((main_x, main_y), product_id, fill=TEXT_COLOR, font=font_main)

            # ===== 2行目（名前） =====
            # 12文字以上の場合は改行（自動で適切に折り返し）
            lines = textwrap.wrap(name, width=12)
            text_sub = "\n".join(lines)
            w_sub, h_sub = draw.multiline_textsize(text_sub, font=font_sub, spacing=10)
            sub_x = (IMG_SIZE[0] - w_sub) / 2
            sub_y = IMG_SIZE[1] * 2 / 3 - (h_sub / 2)
            draw.multiline_text((sub_x, sub_y), text_sub, fill=TEXT_COLOR, font=font_sub, align="center", spacing=10)

            # 保存
            img.save(output_path, "JPEG")
            print(f"生成: {output_path}")

print("完了しました。")
```
