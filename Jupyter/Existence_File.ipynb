```python
import pandas as pd
import os

# --- 1. 設定項目 (ご自身の環境に合わせて変更してください) ---

# 入力するCSVファイルのパス
input_csv_path = '../data/csv/Data_20251007.csv' 

# 画像ファイルが格納されているディレクトリのパス
image_directory = '../images/'

# 結果を出力する新しいCSVファイルのパス
output_csv_path = 'Existence_Check_result.csv'

# チェック対象の列名 (CSVファイルのヘッダーに合わせてください)
column_name = 'img_file_Name'

# ----------------------------------------------------------

print(f"'{input_csv_path}' を読み込んで処理を開始します...")

# --- 2. メイン処理 ---
try:
    # CSVファイルをデータフレームとして読み込む
    df = pd.read_csv(input_csv_path)

    # 存在チェックの結果を格納するリストを作成
    existence_results = []

    # 対象列の各ファイル名についてループ処理
    for filename in df[column_name]:
        # 画像ファイルへのフルパスを作成
        # (str(filename)で、万が一数値として読み込まれても文字列に変換)
        full_path = os.path.join(image_directory, str(filename))
        
        # ファイルが存在するかチェック
        if os.path.exists(full_path):
            existence_results.append('○')
        else:
            existence_results.append('×')

    # --- 3. 結果の書き出し ---
    
    # 新しいデータフレームを作成 (元のファイル名列と、存在チェック結果の列)
    result_df = pd.DataFrame({
        column_name: df[column_name],
        '存在チェック': existence_results
    })

    # 新しいCSVファイルとして書き出す
    # index=False: 行番号は不要なため出力しない
    # encoding='utf-8-sig': Excelで開いた際の文字化けを防ぐ
    result_df.to_csv(output_csv_path, index=False, encoding='utf-8-sig')

    print("-" * 30)
    print(f"処理が完了しました！")
    print(f"結果は '{output_csv_path}' に保存されました。")
    
    # 結果の先頭5行を表示して確認
    print("\n▼ 出力ファイルのプレビュー:")
    display(result_df.head())

except FileNotFoundError:
    print(f"エラー: 入力ファイル '{input_csv_path}' が見つかりません。パスを確認してください。")
except KeyError:
    print(f"エラー: CSVファイルに '{column_name}' という列が見つかりません。列名を確認してください。")
```
